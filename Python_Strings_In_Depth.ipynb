## 1. What is a string?

A **string** is a sequence of Unicode characters. Python uses the type `str` for text.

Strings may contain letters, numbers, whitespace, punctuation, and characters from many writing systems.

Remember the difference between a number and a string containing digits:
- `123` is an integer.
- `"123"` is a string.

```python
name = "Rahul"
number_as_text = "123"
number = 123
multilingual = "Hello"

print(name, type(name))
print(number_as_text, type(number_as_text))
print(number, type(number))
print(multilingual)
```

### Creating strings

Single quotes, double quotes, and triple quotes can all be used. Triple-quoted strings are useful for multiline text.

`str(value)` converts a value to its string representation. An empty string `""` has length zero and is falsy.

```python
single = 'Hello'
double = "Python"
multiline = """This is
a multiline
string."""

print(single)
print(double)
print(multiline)
print(str(25), type(str(25)))
print(len(""), bool(""))
```

## 2. Quotes, escape sequences, and raw strings

Common escape sequences:

| Sequence | Meaning |
|---|---|
| `\\n` | New line |
| `\\t` | Tab |
| `\\\\` | Backslash |
| `\\'` | Single quote |
| `\\"` | Double quote |
| `\\r` | Carriage return |
| `\\u03A9` | Unicode code point Ω |

A raw string uses the `r` prefix and is useful for Windows paths and regular expressions. A raw string cannot end with an odd number of backslashes.

```python
print("First line\nSecond line")
print("Name:\tAman")
print("C:\\Users\\Student")
print('It\'s Python')
print("\u03A9")

path = r"C:\Users\Student\Documents"
print(path)
```

## 3. Indexing

Indexing retrieves one character. Positive indices begin at zero; negative indices count from the end.

For `"PYTHON"`:
- `text[0]` is `"P"`
- `text[2]` is `"T"`
- `text[-1]` is `"N"`
- `text[-2]` is `"O"`

An invalid index raises `IndexError`.

```python
text = "PYTHON"

print(text[0])
print(text[2])
print(text[-1])
print(text[-2])

index = 10
if -len(text) <= index < len(text):
    print(text[index])
else:
    print("Index out of range")
```

## 4. Slicing

Syntax: `string[start:stop:step]`

- `start` is included.
- `stop` is excluded.
- `step` determines the gap between selected positions.
- Omitting boundaries uses defaults.
- A negative step can reverse the traversal direction.

Unlike indexing, ordinary out-of-range slice boundaries do not raise `IndexError`.

```python
text = "ABCDEFGHIJ"

print(text[0:3])     # ABC
print(text[2:5])     # CDE
print(text[:4])      # ABCD
print(text[4:])      # EFGHIJ
print(text[::2])     # ACEGI
print(text[1::2])    # BDFHJ
print(text[::-1])    # JIHGFEDCBA
print(text[-4:-1])   # GHI
print(text[100:200]) # empty string
```

## 5. String immutability

Strings are **immutable**: you cannot change an individual character in place. String methods generally return a new string.

The following cell demonstrates the correct pattern.

```python
text = "Python"
# text[0] = "J"  # Uncomment to see TypeError

text = "J" + text[1:]
print(text)

original = "python"
converted = original.upper()
print(original)   # unchanged
print(converted)  # new string
```

## 6. String operators

- `+`: concatenate strings
- `*`: repeat a string
- `in` / `not in`: substring membership
- `==`, `!=`, `<`, `>`: compare strings
- Comparisons are case-sensitive and use lexicographical Unicode code-point ordering.

Use `str()` or an f-string when combining text with numbers.

```python
print("Hello" + " " + "Python")
print("Hi! " * 3)

text = "Python Programming"
print("Python" in text)
print("Java" not in text)
print("Python" == "python")
print("apple" < "banana")

age = 25
print("Age: " + str(age))
print(f"Age: {age}")
```

## 7. Case conversion methods

| Method | Purpose |
|---|---|
| `upper()` | Uppercase applicable characters |
| `lower()` | Lowercase applicable characters |
| `capitalize()` | Uppercase first character, lowercase the rest |
| `title()` | Title-case transformation using Python's word rules |
| `swapcase()` | Swap the case of cased characters |
| `casefold()` | Aggressive case normalization for caseless matching |

`title()` is not a perfect natural-language title formatter. `casefold()` is often preferable to `lower()` for Unicode-aware caseless comparison.

```python
text = "pYTHON PROGRAMMING"

print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.title())
print("PyThOn".swapcase())

a = "Straße"
b = "STRASSE"
print(a.casefold() == b.casefold())
```

## 8. Searching and counting

- `find(sub)`: lowest matching index, or `-1` if absent.
- `rfind(sub)`: highest matching index, or `-1` if absent.
- `index(sub)`: like `find()`, but raises `ValueError` if absent.
- `rindex(sub)`: rightmost match, but raises `ValueError` if absent.
- `count(sub)`: counts non-overlapping occurrences.

These methods can accept optional `start` and `end` boundaries.

```python
text = "banana"

print(text.find("a"))
print(text.rfind("a"))
print(text.index("na"))
print(text.rindex("na"))
print(text.count("a"))
print(text.count("ana"))   # non-overlapping occurrences
print(text.find("z"))      # -1
print("abc".count(""))     # 4
```

## 9. Replacing and translating

`replace(old, new, count)` returns a new string with replacements. The optional `count` limits how many replacements occur.

`str.maketrans()` and `translate()` support character-by-character translation and removal.

```python
text = "Java is useful. Java is popular."
print(text.replace("Java", "Python"))
print(text.replace("Java", "Python", 1))

table = str.maketrans("aeiou", "12345")
print("education".translate(table))

remove_vowels = str.maketrans("", "", "aeiouAEIOU")
print("Hello World".translate(remove_vowels))
```

## 10. Splitting, joining, and trimming

- `split()` splits into a list of strings.
- `rsplit()` splits from the right when a maximum split count is supplied.
- `splitlines()` splits at line boundaries.
- `separator.join(iterable)` joins strings.
- `strip()`, `lstrip()`, and `rstrip()` remove characters from the ends.

**Important:** `split()` without a separator handles runs of whitespace differently from `split(",")`. The argument to `strip(chars)` is a set of characters, not an exact substring.

```python
print("Python is easy".split())
print("apple,banana,mango".split(","))
print("a-b-c-d".split("-", 2))
print("a-b-c-d".rsplit("-", 2))
print("First\nSecond\nThird".splitlines())

words = ["Python", "is", "powerful"]
print(" ".join(words))
print("-".join(words))

print("   Python   ".strip())
print("   Python   ".lstrip())
print("   Python   ".rstrip())
print("xxHelloxx".strip("x"))
print("Mr. Pradyumna".removeprefix("Mr. "))
print("report.csv".removesuffix(".csv"))
```

## 11. String validation methods

These methods return Boolean values and generally return `False` for an empty string, except `isascii()` and some other methods with their own definitions.

| Method | Checks |
|---|---|
| `isalpha()` | All characters alphabetic |
| `isalnum()` | All characters alphabetic or numeric |
| `isascii()` | All characters ASCII |
| `isdecimal()` | Decimal characters |
| `isdigit()` | Digit characters, including some superscripts |
| `isnumeric()` | A broader range of numeric characters |
| `islower()` / `isupper()` | Cased characters meet the case condition |
| `istitle()` | Title-case pattern |
| `isspace()` | All characters whitespace |
| `isprintable()` | All characters printable |
| `isidentifier()` | Valid Python identifier syntax |

A string can pass `isidentifier()` and still be a reserved keyword, such as `"class"`. Check keywords with `keyword.iskeyword()`.

```python
import keyword

examples = ["Python", "Python123", "Python 123", "123", "²", "½", "   ", "student_name", "class"]

for item in examples:
    print(
        repr(item),
        "alpha:", item.isalpha(),
        "alnum:", item.isalnum(),
        "decimal:", item.isdecimal(),
        "digit:", item.isdigit(),
        "numeric:", item.isnumeric(),
        "identifier:", item.isidentifier()
    )

print(keyword.iskeyword("class"))
print("Hello".isupper())
print("HELLO".isupper())
print("   ".isspace())
print("Hello\nWorld".isprintable())
```

## 12. Prefixes, suffixes, and alignment

- `startswith()` / `endswith()` check prefixes and suffixes.
- They can accept a tuple of alternatives.
- `center()`, `ljust()`, and `rjust()` align text in a specified width.
- `zfill()` pads on the left with zeros and handles a leading sign.

These methods operate on string positions, not necessarily the visual width of Unicode characters in a terminal.

```python
filename = "photo.png"
print(filename.startswith("photo"))
print(filename.endswith((".jpg", ".png", ".gif")))

print("Python".center(12, "-"))
print("Python".ljust(10, "-"))
print("Python".rjust(10, "-"))
print("42".zfill(5))
print("-42".zfill(5))
```

## 13. String formatting

F-strings are usually the clearest modern option. `str.format()` and older percent formatting are also supported.

Formatting examples:
- `{value:.2f}`: two digits after the decimal point
- `{value:,.2f}`: thousands separators and two decimal places
- `{value:.1%}`: percentage formatting
- `>`, `<`, `^`: right, left, and center alignment

```python
name = "Aman"
age = 20
price = 1234.5678
ratio = 0.25

print(f"My name is {name} and I am {age}.")
print(f"Sum = {10 + 20}")
print(f"{price:.2f}")
print(f"{price:,.2f}")
print(f"{ratio:.1%}")
print(f"{'Python':>10}")
print(f"{'Python':<10}")
print(f"{'Python':^10}")

print("Hello, {}!".format(name))
print("{0} teaches {1}.".format(name, "Python"))
print("{name} is {age} years old.".format(name=name, age=age))

data = {"name": "Aman", "age": 20}
print("{name} is {age} years old.".format_map(data))

print("Student: %s, Score: %d" % (name, 95))
```

## 14. Iterating over strings

A `for` loop can visit each character. `enumerate()` also provides an index.

List comprehensions are shown here as a string-processing technique; Lists themselves will be studied in the next module.

```python
text = "Python"

for character in text:
    print(character)

for index, character in enumerate(text):
    print(index, character)

characters = [character.upper() for character in text]
digits = [character for character in "Python123" if character.isdigit()]

print(characters)
print(digits)
```

## 15. Built-in functions used with strings

Useful functions include:
- `len()`, `str()`, `repr()`, `type()`
- `ord()` and `chr()`
- `min()`, `max()`, `sorted()`
- `reversed()`, `enumerate()`
- `any()` and `all()`

`ord()` accepts a string containing exactly one Unicode code point. `sorted()` returns a list of characters.

```python
text = "Python"
print(len(text))
print(repr("line\nbreak"))
print(ord("A"))
print(chr(65))
print(min("cab"))
print(max("cab"))
print(sorted("cab"))
print("".join(reversed(text)))
print(any(ch.isdigit() for ch in "a1b"))
print(all(ch.isalpha() for ch in "Python"))
```

## 16. Unicode, encoding, and decoding

Python `str` represents Unicode text. Encoding converts text into bytes; decoding converts bytes back into text.

- `str`: text
- `bytes`: byte values
- UTF-8: widely used text encoding

A Unicode code point is not necessarily the same as the number of bytes used to encode a character.

```python
text = "Hello नमस्ते Ω"
print(text)
print("Code points in string:", len(text))

encoded = "नमस्ते".encode("utf-8")
print(encoded)
print(type(encoded))
print(encoded.decode("utf-8"))

print(ord("Ω"), chr(ord("Ω")))
print("A".encode("utf-8"))
print("Ω".encode("utf-8"))
```

## 17. Regular expressions

The `re` module supports pattern-based searching, extraction, replacement, and splitting.

Common patterns:
- `\d+`: one or more digit characters
- `\s+`: one or more whitespace characters
- `[A-Z]`: an uppercase ASCII letter
- `.`: any character except newline by default

Regular expressions are powerful but do not automatically provide perfect validation for complex formats such as email addresses.

```python
import re

text = "My numbers are 9876543210 and 12345."
print(re.search(r"\d+", text).group())
print(re.findall(r"\d+", text))
print(re.sub(r"\s+", " ", "Python    is   easy"))
print(re.split(r"[,; ]+", "apple,banana;orange mango"))

# Extract a simple extension from a filename (not a replacement for pathlib)
filename = "report.csv"
match = re.search(r"\.([A-Za-z0-9]+)$", filename)
print(match.group(1) if match else "No extension")
```

## 18. Common mistakes

1. Trying to assign to `text[0]` — strings are immutable.
2. Confusing `find()` (`-1` if absent) with `index()` (`ValueError` if absent).
3. Treating `strip("abc")` as removal of the exact substring `"abc"`.
4. Forgetting that `split()` returns a list.
5. Writing `text.upper` instead of calling `text.upper()`.
6. Expecting `text.upper()` to modify the original string.
7. Treating `isnumeric()`, `isdigit()`, and `isdecimal()` as identical.
8. Assuming all string operations are case-insensitive.
9. Concatenating a string and an integer without conversion.
10. Assuming every Unicode character occupies one byte.

```python
text = "python"
print(text.upper)    # bound method object; not the converted text
print(text.upper())  # actual result

print("Python".find("z"))  # -1
# "Python".index("z")     # ValueError if uncommented

print("a,,b".split(","))
print("a  b".split())
print("²".isdecimal(), "²".isdigit(), "½".isnumeric())
```

## 20. Guided solutions for selected questions

Try solving each problem yourself first. These examples demonstrate one possible approach; many problems have multiple valid solutions.

```python
# 1. Palindrome (case-insensitive; ignores surrounding whitespace)
def is_palindrome(text):
    normalized = text.strip().casefold()
    return normalized == normalized[::-1]

print(is_palindrome("Level"))
print(is_palindrome("Python"))

# 2. Count vowels
def count_vowels(text):
    vowels = "aeiou"
    return sum(1 for ch in text.casefold() if ch in vowels)

print(count_vowels("Python Programming"))

# 3. Character frequency (counts exact characters)
def character_frequency(text):
    counts = {}
    for ch in text:
        counts[ch] = counts.get(ch, 0) + 1
    return counts

print(character_frequency("banana"))

# 4. Remove duplicate characters while preserving first occurrence
def unique_characters(text):
    result = ""
    for ch in text:
        if ch not in result:
            result += ch
    return result

print(unique_characters("banana"))

# 5. Normalize repeated whitespace
import re
def normalize_whitespace(text):
    return re.sub(r"\s+", " ", text).strip()

print(normalize_whitespace("  Python   is\n\t powerful  "))
```
