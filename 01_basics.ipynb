3. Defining Variables

```python
# 3. VARIABLES
name = "Rahul"
age = 21
height = 5.8
is_student = True

print(name)
print(age)
print(height)
print(is_student)
```

4. Defining Multiple Variables

```python
# 4. MULTIPLE VARIABLE ASSIGNMENT
a, b, c = 10, 20, 30

print(a, b, c)

x = y = z = 100
print(x, y, z)
```

5. Checking Data types

```python
# 5. DATA TYPES
integer_value = 100
float_value = 10.5
string_value = "Python"
boolean_value = True
none_value = None

print(type(integer_value))
print(type(float_value))
print(type(string_value))
print(type(boolean_value))
print(type(none_value))
```

6. Type Casting

```python
# 6. TYPE CASTING
number = "100"

converted_number = int(number)

print(converted_number)
print(type(converted_number))

price = 99
converted_price = float(price)

print(converted_price)

age = 21
age_text = str(age)

print(age_text)
print(type(age_text))
```

8. Output Formatting

```python
# 8. BASIC OUTPUT FORMATTING
name = "Rahul"
age = 21
city = "Delhi"

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"I live in {city}.")
```

9. Variable Naming Rule

```python
# 10. VARIABLE NAMING RULES
student_name = "Amit"
total_marks = 450
_subject = "Python"

print(student_name, total_marks, _subject)

# Valid: student_name, age1, _value
# Invalid: 1age, student-name, class
```

11. SWAPPING VARIABLES

```python
# 11. SWAPPING VARIABLES
a = 10
b = 20

a, b = b, a
print("After swapping:")
print("a =", a)
print("b =", b)
```
