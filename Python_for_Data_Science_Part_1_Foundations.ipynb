## 3. Variables and assignment

A variable is a name that refers to an object/value. Python uses `=` for assignment.

### Naming rules
- Names may contain letters, digits, and underscores.
- A name cannot begin with a digit.
- Names are case-sensitive: `age` and `Age` are different.
- Python keywords such as `if`, `for`, and `class` cannot be used as ordinary variable names.
- Prefer descriptive `snake_case` names.

````python
student_name = "Aarav"
age = 22
height_m = 1.72
is_enrolled = True

print(student_name)
print(age)
print(height_m)
print(is_enrolled)
````

````python
# Multiple assignment
x, y, z = 10, 20, 30
print(x, y, z)

# Updating a variable
score = 75
score = score + 5
print("Updated score:", score)

# Swap two variables
x, y = y, x
print("After swapping:", x, y)
````

## 4. Core data types

| Type | Example | Typical use |
|---|---|---|
| `int` | `25` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `str` | `"Python"` | Text |
| `bool` | `True` | Logical truth values |
| `NoneType` | `None` | Represents absence of a value |

Use `type(value)` to inspect a value's type. Python is dynamically typed: a name can later refer to a value of another type.

````python
items = [42, 3.14, "data", True, None]

for item in items:
    print(item, "->", type(item))
````

### Type conversion (casting)

Conversions should be intentional. For example, `int("12")` works, but `int("12.5")` raises a `ValueError`.

````python
age_text = "24"
age_number = int(age_text)

price_text = "199.50"
price = float(price_text)

quantity = 3
message = str(quantity) + " items"

print(age_number, type(age_number))
print(price, type(price))
print(message)
````

## 5. Arithmetic operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `8 + 2` |
| `-` | Subtraction | `8 - 2` |
| `*` | Multiplication | `8 * 2` |
| `/` | True division | `7 / 2` gives `3.5` |
| `//` | Floor division | `7 // 2` gives `3` |
| `%` | Remainder | `7 % 2` gives `1` |
| `**` | Exponentiation | `2 ** 3` gives `8` |

Remember that `//` rounds down toward negative infinity, which matters for negative values.

````python
a, b = 17, 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("True division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** 2)

print("Negative floor division:", -7 // 2)
````

## 6. Comparison and logical operators

Comparison operators return a Boolean value: `==`, `!=`, `<`, `<=`, `>`, `>=`.

Logical operators combine conditions:
- `and`: both conditions must be true.
- `or`: at least one condition must be true.
- `not`: reverses a Boolean result.

Use `==` to compare values and `=` to assign a value.

````python
age = 24
has_id = True

print(age >= 18)
print(age == 24)
print(age != 21)
print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
````

### Operator precedence

Parentheses make complex expressions easier to read. When in doubt, use parentheses rather than relying on memory of precedence.

````python
result = (5 + 3) * 2
print(result)

is_eligible = (age >= 18) and has_id
print(is_eligible)
````

## 7. Strings

A string (`str`) is an immutable sequence of Unicode characters. Strings can use single quotes, double quotes, or triple quotes.

````python
single = 'Python'
double = "Data Science"
multiline = '''Line one
Line two'''

print(single)
print(double)
print(multiline)
````

### Indexing and slicing

Python uses zero-based indexing. Negative indexes count from the end. The stop position in a slice is excluded.

````python
language = "Python"

print(language[0])     # P
print(language[-1])    # n
print(language[0:3])   # Pyt
print(language[2:])    # thon
print(language[:4])    # Pyth
print(language[::-1]) # nohtyP
````

### Useful string methods

Strings are immutable, so methods such as `replace()` return a new string rather than changing the original.

````python
raw_text = "  machine learning with Python  "

print(raw_text.strip())
print(raw_text.lower())
print(raw_text.upper())
print(raw_text.title())
print(raw_text.replace("Python", "SQL"))
print("a,b,c".split(","))
print("-".join(["2026", "10", "13"]))
print("Python".startswith("Py"))
print("learning" in raw_text)
````

### String formatting with f-strings

F-strings are a clear way to embed values in text. Format specifiers can control decimal places and alignment.

````python
name = "Meera"
marks = 87.456
print(f"{name} scored {marks:.2f}%")

product = "Notebook"
price = 49.5
quantity = 4
print(f"{quantity} {product}(s) cost ₹{price * quantity:.2f}")
````

### Common string pitfalls
- Strings cannot be modified in place: `text[0] = "P"` raises `TypeError`.
- `len(text)` counts characters, not words.
- `split()` without an argument splits on runs of whitespace.

````python
text = "Python is useful for data analysis"
words = text.split()
print("Characters:", len(text))
print("Words:", len(words))
print("Words list:", words)
````

## 8. Date and time values

Python's standard-library `datetime` module supports dates, times, and date-time values. Use explicit formats when parsing dates from text.

````python
from datetime import date, datetime, time, timedelta

today = date.today()
now = datetime.now()
class_start = time(10, 30)

print("Today:", today)
print("Current local date-time:", now)
print("Class start:", class_start)
````

````python
# Creating and formatting a date
date_text = "2026-10-13"
parsed_date = datetime.strptime(date_text, "%Y-%m-%d").date()

print(parsed_date)
print(parsed_date.strftime("%d %B %Y"))
print(parsed_date.strftime("%A"))
````

````python
start_date = date(2026, 10, 1)
end_date = date(2026, 10, 13)
difference = end_date - start_date

print("Days between dates:", difference.days)
print("One week later:", start_date + timedelta(days=7))
````

### Common format codes

| Code | Meaning | Example |
|---|---|---|
| `%Y` | Four-digit year | `2026` |
| `%m` | Month number | `10` |
| `%d` | Day of month | `13` |
| `%H` | Hour, 24-hour clock | `19` |
| `%M` | Minute | `25` |
| `%S` | Second | `00` |
| `%A` | Full weekday name | `Tuesday` |
| `%B` | Full month name | `October` |

A date format mismatch can raise `ValueError`. For timestamps from different time zones, use timezone-aware datetimes rather than assuming all times are local.

## 10. A small applied example: simple bill calculator

This example combines variables, type conversion, arithmetic, and formatted output.

````python
item_name = "Notebook"
unit_price = 49.50
quantity = 4
discount_rate = 0.10

subtotal = unit_price * quantity
discount = subtotal * discount_rate
total = subtotal - discount

print("----- BILL -----")
print(f"Item: {item_name}")
print(f"Quantity: {quantity}")
print(f"Subtotal: ₹{subtotal:.2f}")
print(f"Discount: ₹{discount:.2f}")
print(f"Final total: ₹{total:.2f}")
````
