1.  ARITHMETIC OPERATORS

```python
# 1. ARITHMETIC OPERATORS

a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# 2. COMPARISON OPERATORS

x = 10
y = 20

print(x == y)  # Equal to
print(x != y)  # Not equal to
print(x > y)   # Greater than
print(x < y)   # Less than
print(x >= y)  # Greater than or equal to
print(x <= y)  # Less than or equal to
```

```python
# 3. LOGICAL OPERATORS

age = 22
has_id = True

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
```

```python
# 4. ASSIGNMENT OPERATORS

number = 10

number += 5
print(number)

number -= 3
print(number)

number *= 2
print(number)

number /= 4
print(number)
```

```python
# 5. MEMBERSHIP OPERATORS

language = "Python"

print("P" in language)
print("z" not in language)
```

```python
# 6. IDENTITY OPERATORS

a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a == c)
```

```python
# 7. OPERATOR PRECEDENCE

result = 10 + 5 * 2
print(result)

result = (10 + 5) * 2
print(result)
```

```python
# 8. CREATING STRINGS

name = "Python"
message = 'Welcome to Python'

print(name)
print(message)
```

```python
# 9. STRING LENGTH

text = "Programming"

print(len(text))
```

```python
# 10. STRING INDEXING

language = "Python"

print(language[0])
print(language[1])
print(language[-1])
print(language[-2])
```

```python
# 11. STRING SLICING

language = "Python Programming"

print(language[0:6])
print(language[7:18])
print(language[:6])
print(language[7:])
print(language[-11:])
print(language[::2])
print(language[::-1])
```

```python
# 12. STRING CASE METHODS

text = "Hello Python"

print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
print(text.swapcase())
```

```python
# 13. REMOVING WHITESPACE

text = "   Python Programming   "

print(text.strip())
print(text.lstrip())
print(text.rstrip())
```

```python
# 14. FINDING AND COUNTING

text = "Python is easy. Python is powerful."

print(text.find("Python"))
print(text.count("Python"))
print(text.index("easy"))
```

```python
# 15. REPLACING STRINGS

text = "I like Java"

new_text = text.replace("Java", "Python")

print(new_text)
```

```python
# 16. SPLIT AND JOIN

text = "Python,Java,C++"

languages = text.split(",")

print(languages)

result = " | ".join(languages)

print(result)
```

```python
# 17. CHECKING STRING CONTENT

print("python".isalpha())
print("12345".isdigit())
print("Python123".isalnum())
print("hello".islower())
print("HELLO".isupper())
print("   ".isspace())
```

```python
# 18. STRING STARTS WITH AND ENDS WITH

filename = "student_data.csv"

print(filename.startswith("student"))
print(filename.endswith(".csv"))
```

```python
# 19. STRING FORMATTING USING F-STRINGS

name = "Rahul"
age = 21
marks = 89.567

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"My marks are {marks:.2f}.")
```

```python
# 20. ESCAPE CHARACTERS

print("Hello\nWorld")
print("Hello\tPython")
print("He said, \"Python is easy\"")
```
